```python
def maxindex(d):
    """Zwraca indeks najniższego niezerowego elementu (od końca)."""
    for i in range(len(d) - 1, -1, -1):
        if abs(d[i]) > 1e-9:  # Tolerancja dla ułamków
            return i
    return -1

def remove_pivot_rows_original(col, T, max_steps=10):
    """Wersja Zomorodian & Carlsson (2005) - BŁĘDNA dla ciał innych niż Z_2"""
    d = list(col)  # Kopiujemy kolumnę wejściową
    print(f"  Start: d = {d}")
    
    steps = 0
    while True:
        steps += 1
        if steps > max_steps:
            print("  [!] PRZERWANO: Wykryto nieskończoną pętlę! Kolumna nigdy się nie wyzeruje.")
            break
            
        i = maxindex(d)
        if i == -1:  # d jest puste (same zera)
            break
        if T[i] is None:
            break
            
        # Linijka 7 z oryginalnego pseudokodu: q = coefficient of sigma^i in T[i]
        q = T[i][i]
        
        # Linijka 8 z oryginalnego pseudokodu: d = d - q^{-1} T[i]
        for r in range(len(d)):
            d[r] = d[r] - (1.0 / q) * T[i][r]
            
        print(f"  Krok {steps}: d = {d}")
        
    return d

def remove_pivot_rows_corrected(col, T, max_steps=10):
    """Wersja poprawna algebraicznie - z uwzględnieniem c"""
    d = list(col)
    print(f"  Start: d = {d}")
    
    steps = 0
    while True:
        steps += 1
        if steps > max_steps:
            break
            
        i = maxindex(d)
        if i == -1:
            break
        if T[i] is None:
            break
            
        # Poprawka: pobieramy też c z kolumny d
        q = T[i][i]
        c = d[i]
        
        # Zaktualizowany wzór: d = d - c * q^{-1} T[i]
        for r in range(len(d)):
            d[r] = d[r] - c * (1.0 / q) * T[i][r]
            
        print(f"  Krok {steps}: d = {d}")
        
    return d

# ==========================================
# URUCHOMIENIE TESTU
# ==========================================

# Symulujemy stan tablicy T.
# Na indeksie 1 mamy kolumnę [0, 2, 0] (pivot = 2)
T = [None, [0.0, 2.0, 0.0], None]

# Nowa kolumna d, którą chcemy zredukować
nowa_kolumna = [0.0, 6.0, 0.0]

print("=== TEST 1: Oryginalny algorytm z 2005 (bez mnożenia przez c) ===")
wynik_orig = remove_pivot_rows_original(nowa_kolumna, T)

print("\n=== TEST 2: Poprawiony algorytm (z uwzględnieniem c) ===")
wynik_corr = remove_pivot_rows_corrected(nowa_kolumna, T)
```
